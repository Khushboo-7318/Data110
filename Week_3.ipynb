```python
import matplotlib.pyplot as plt
```

```python
study_hours = [1, 2, 2.5, 3, 4, 5, 5.5, 6, 7, 8]
exam_scores = [55, 60, 63, 66, 72, 78, 80, 84, 88, 94]

plt.scatter(study_hours, exam_scores, color="blue", marker="o")
plt.title("Study Time vs. Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.grid(True)
plt.show()
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
city_a = [68, 70, 72, 71, 74, 76, 75]
city_b = [62, 64, 67, 66, 69, 71, 70]
plt.plot(days, city_a, marker="o", linestyle="-", label="City A")
plt.plot(days, city_b, marker="s", linestyle="--", label="City B")
plt.title("Temperature Over One Week")
plt.xlabel("Day")
plt.ylabel("Temperature (°F)")
plt.legend()
plt.grid(True)
plt.show()
```

City A was generally warmer during the week.

```python
offenses = ["Assault", "Burglary", "Larceny", "Vehicle Theft",
            "Murder", "Rape", "Robbery"]

crime_rates = [264.1, 250.7, 1347.2, 318.7, 5.7, 38.0, 66.5]
plt.figure(figsize=(10, 5))
plt.bar(offenses, crime_rates, color="skyblue")
plt.title("Crime Rates in 2023")
plt.xlabel("Type of Crime")
plt.ylabel("Offenses per 100,000 People")
plt.grid(axis="y")
plt.show()
```

```python
changes = [-3.3, -8.1, -4.9, 12.0, -12.0, -9.8, -0.8]
colors = ["red", "orange", "blue", "green", "purple", "brown", "pink"]
plt.figure(figsize=(10, 5))
plt.barh(offenses, changes, color=colors)
plt.title("Change in Crime Rates: 2022–2023")
plt.xlabel("Change per 100,000 People")
plt.ylabel("Type of Crime")
plt.axvline(0, color="black")
plt.grid(axis="x")
plt.show()
```

This graph shows which crimes increased or decreased from 2022 to 2023. It helps us understand how crime rates changed.

```python
plt.figure(figsize=(10, 6))
colors = ["red" if x < 0 else "green" for x in changes]
plt.barh(offenses, changes, color=colors)
plt.title("Improved Crime Rate Changes (2022–2023)", fontsize=14)
plt.xlabel("Change per 100,000 People")
plt.axvline(0, color="black")
plt.grid(axis="x", linestyle="--")
plt.show()
```

Red means crime rates decreased. Green means crime rates increased.

```python
import numpy as np

companies = ["Company A", "Company B", "Company C"]
product1 = [20, 34, 30]
product2 = [25, 32, 34]
product3 = [30, 35, 27]

x = np.arange(3)
width = 0.25

plt.figure(figsize=(8, 5))

plt.bar(x - width, product1, width, label="Product 1")
plt.bar(x, product2, width, label="Product 2")
plt.bar(x + width, product3, width, label="Product 3")
plt.title("Product Sales by Company")
plt.xlabel("Company")
plt.ylabel("Sales")
plt.xticks(x, companies)
plt.legend()
plt.grid(axis="y")
plt.show()
```
